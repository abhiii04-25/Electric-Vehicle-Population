### 8. Which manufacturer has the most EVs?

```python
top_manufacturer=df["Make"].value_counts().idxmax()
count=df["Make"].value_counts().max()
print("Manufacturer with most EV's: ",top_manufacturer)
print("Number of EV's: ",count)
```

### 9. Which model appears the most?

```python
most_common=df["Model"].value_counts().idxmax()
count=df["Model"].value_counts().max()
print("Most common Models are: ",most_common)
print("Count of the models: ",count)
```

### 10. How many different EV models are there?

```python
models=sorted(df["Model"].unique())
print("Numbers of Models: ",models)
```
