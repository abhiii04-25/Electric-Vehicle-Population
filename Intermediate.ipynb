# Pandas + Visualization

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

### Top 10 EV manufacturers.

```python
df=pd.read_csv("EV.csv")
df.head()
```

```python
top10=df["Make"].value_counts().head(10)
plt.figure(figsize=(10, 6))
top10.plot(kind="bar")

plt.title("Top 10 EV Manufacturers")
plt.xlabel("Number of EVs")
plt.ylabel("Manufacturer")
plt.show()
```

### Top 10 EV models.

```python
top10EV=df["Model"].value_counts().head(10)
plt.figure(figsize=(10, 6))
top10EV.plot(kind="bar")

plt.title("Top 10 EV Models")
plt.xlabel("Number of EVs")
plt.ylabel("Manufacturer")
plt.show()
```

### Distribution of EV model years

```python
years=df["Model Year"].value_counts().head()

plt.figure(figsize=(12, 6))
sns.barplot(x=years.index, y=years.values,palette="rainbow")

plt.title("Distribution of EV Model Years")
plt.xlabel("Model Year")
plt.ylabel("Number of EVs")
plt.show()
```

### 

### Which year had the highest number of registrations?

```python
year_counts = df["Model Year"].value_counts()

print("Year with highest registrations:", year_counts.idxmax())
print("Number of registrations:", year_counts.max())
```

### Average electric range by manufacturer.

```python
Range=df.groupby("Make")["Electric Range"].mean().sort_values(ascending=False).head(10)
print("Average electric range by manufacturer.",Range)
sns.barplot(y=Range.index,x=Range.values)
plt.title("Top 10 Manufacturers by Average Electric Range")
plt.xlabel("Manufacturer")
plt.ylabel("Average Electric Range")
```

### Maximum electric range by model.

```python
MaxRange=df.groupby("Model")["Electric Range"].max().sort_values(ascending=False).head(10)
sns.barplot(x=MaxRange.values, y=MaxRange.index)

plt.title("Top 10 EV Models by Maximum Electric Range")
plt.xlabel("Maximum Electric Range")
plt.ylabel("Model")
plt.show()
```

### Top 15 cities with the highest EV population

```python
top15 = df["City"].value_counts().head(15)

plt.figure(figsize=(12, 6))
sns.barplot(y=top15.index, x=top15.values)

plt.title("Top 15 Cities with the Highest EV Population")
plt.xlabel("City")
plt.ylabel("Number of EVs")

```

### Top counties by EV registrations

```python

top15 = df["County"].value_counts().head(15).index

sns.countplot(
    data=df,
    y="County",
    order=top15
)

plt.title("Top 15 Counties by EV Registrations")
plt.xlabel("County")
plt.ylabel("Number of EV Registrations")
```

### Number of Battery Electric Vehicles (BEV) vs Plug-in Hybrid Electric Vehicles (PHEV)

```python
vehicle_counts = df["Electric Vehicle Type"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(vehicle_counts,labels=vehicle_counts.index,autopct="%1.1f%%",startangle=90)
plt.title("BEV vs PHEV Distribution")
plt.show()
```

### Percentage of each EV type

```python
percentage = df["Electric Vehicle Type"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(percentage,labels=percentage.index,autopct="%1.1f%%",startangle=90)
plt.title("Percentage of Each EV Type")
plt.show()
```
