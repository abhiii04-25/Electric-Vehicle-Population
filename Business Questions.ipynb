```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns  
```

## Imagine you're working for an EV company:

### 1 Which city should Tesla expand into?

```python
df=pd.read_csv("EV.csv")
tesla = df[df['Make'] == 'TESLA']
tesla_city = tesla['City'].value_counts().head(10)
print(tesla_city)
```

```python
plt.figure(figsize=(10,6))
tesla_city.plot(kind='bar', color='orange')

plt.title('Top Cities with Tesla Vehicles')
plt.xlabel('City')
plt.ylabel('Tesla Registrations')
```

### Average electric range

```python
range_city = (
    df.groupby('City')['Electric Range'].mean().sort_values(ascending=False).head(10)
)
print(range_city)
```

```python
plt.figure(figsize=(10,6))
range_city.plot(kind='bar')
plt.title('Top Cities by Average Electric Range')
plt.ylabel('Average Electric Range')
plt.tight_layout()
plt.show()
```
