# Data Cleaning Challenges

### Fill missing values appropriately.

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df=pd.read_csv("EV.csv")
df.head()
```

```python
null=df.isnull()
print("There are missing values: ",null)
```

### Remove unnecessary columns

```python
df.drop(columns="Legislative District")
```

### Rename columns to cleaner names

```python
df.rename(columns={"Model Year": "Model_Year",
    "Electric Vehicle Type": "EV_Type",
    "Electric Range": "Electric_Range",
    "Base MSRP": "Base_MSRP",
    "Clean Alternative Fuel Vehicle (CAFV) Eligibility": "CAFV_Eligibility",
    "Vehicle Location": "Vehicle_Location",
    "Electric Utility": "Electric_Utility",
    "Legislative District": "Legislative_District",
    "Postal Code": "Postal_Code",
    "County": "County",
    "State": "State",
    "City": "City",
    "Make": "Make",
    "Model": "Model"})
```

### Convert categorical columns into categories

```python
df[df.select_dtypes(include='object').columns] = df.select_dtypes(include='object').astype('category')
df.info()
```

### Detect outliers in Electric Range.

```python
Q1 = df['Electric Range'].quantile(0.25)
Q3 = df['Electric Range'].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outliers = df[(df['Electric Range'] < lower_bound) |
              (df['Electric Range'] > upper_bound)]
```

```python
print("Number of outliers:", len(outliers))
```

```python
print(outliers[['Electric Range']])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.boxplot(df['Electric Range'], vert=False)
plt.title('Boxplot of Electric Range')
plt.xlabel('Electric Range')
plt.show()
```

### Remove impossible Electric Range values (if any).

```python
print(df['Electric Range'].describe())
print("Minimum:", df['Electric Range'].min())
print("Maximum:", df['Electric Range'].max())
```

```python
impossible = df[(df['Electric Range'] < 0) | (df['Electric Range'] > 1000)]

print(impossible)
print("Number of impossible values:", len(impossible))
```

```python
df = df[(df['Electric Range'] >= 0) & (df['Electric Range'] <= 1000)]
```
